```python
import pandas as pd
import polars as pl
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from pathlib import Path
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
```

```python
## Cargar el dataset
data_path = Path.cwd().parent / 'data' / 'registraduria_database_completo.csv'
registraduriaDataset = pl.read_csv(data_path)

## Asignar tipos de datos a las columnas
registraduriaDataset = registraduriaDataset.with_columns([
    # Fecha
    pl.col("full_date").str.to_date("%Y-%m-%d"),

    # Enteros
    pl.col("travel_month").cast(pl.Int8),
    pl.col("travel_quarter").cast(pl.Int8),
    pl.col("travel_year").cast(pl.Int16),
    pl.col("destination_count").cast(pl.Int8),
    pl.col("advance_booking_days").cast(pl.Int8),
    pl.col("prior_authorizations_required").cast(pl.Int8),
    pl.col("position_level").cast(pl.Int8),        # 2.0 → 2

    # Binarios 0/1 → Boolean
    pl.col("is_contractor").cast(pl.Boolean),
    pl.col("is_fiscal_year_end").cast(pl.Boolean),
    pl.col("is_weekend_departure").cast(pl.Boolean),
    pl.col("is_multi_destination").cast(pl.Boolean),
    pl.col("had_travel_incidents").cast(pl.Boolean),
    pl.col("return_trip_included").cast(pl.Boolean),
    pl.col("requires_per_diem").cast(pl.Boolean),

    # Categoricas con muchas categorias → Categorical
    pl.col("city_origin").cast(pl.Categorical()),
    pl.col("city_destination_main").cast(pl.Categorical()),
    pl.col("employee_position_norm").cast(pl.Categorical()),
    pl.col("purpose").cast(pl.Categorical()),

    # Categoricas con pocas categorias → Enum
    pl.col("transport_mode").cast(pl.Enum(["Aéreo", "Terrestre", "Fluvial", "Mixto"])),
    pl.col("accommodation_type").cast(pl.Enum(["Hotel 4-5 estrellas", "Hotel 3 estrellas", "Hospedaje económico", "Sin alojamiento"])),
    pl.col("destination_security_level").cast(pl.Enum(["Bajo riesgo", "Riesgo medio", "Alto riesgo"])),
    pl.col("destination_region").cast(pl.Enum(["Andina", "Caribe", "Pacífica","Orinoquía", "Amazonía", "Insular"])),
    pl.col("spending_tier").cast(pl.Enum(["Bajo", "Medio-Bajo", "Medio-Alto", "Alto"])),
    pl.col("functional_area").cast(pl.Enum(["Registro Civil", "Identificación", "Organización Electoral","Infraestructura TI", "Jurídica", "Administrativa", "Control Interno"])),
])

registraduriaDataset.collect_schema()
```
