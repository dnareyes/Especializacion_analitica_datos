# Cuestionario

## 10. Cargue la información “datos.csv” y considere las transacciones monetarias (V ALOR > 0). ¿Hay diferencia media entre los ingresos y egresos de las mujeres y hombres? ¿Quién gasta más? (Sugerencia: Use pruebas de hipótesis para la media y diferencias de medias y pruebas no paramétricas tipo Wilcoxon).


```python
import pandas as pd
import scipy.stats as stats
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('C:\\Users\\Diego Reyes\\Documents\\Code & Courses\\Especializacion_analitica_datos\\Estadística para analitica de datos\\data\\raw\\data.csv')
print(df.shape)
display(df.describe())

## Filtrar transaccioens monetarias VALOR > 0

df_filter = df[['SEXO', 'INGRESOS', 'EGRESOS']]
df_clean = df_filter[(df_filter['INGRESOS'] > 0) & (df_filter['EGRESOS'] > 0)]
display(df_clean.shape)
display(df_clean.head(10))

## Estadisticas descriptivas por grupo
display(df_clean.groupby('SEXO')[['INGRESOS', 'EGRESOS']].describe().transpose())
```

```python
## 1. Verificar Supuestos 
## 1.1 Verificar Normalidad usando Shaphiro-Wilk
cols = ['INGRESOS', 'EGRESOS']
grupos = list(df_clean['SEXO'].unique())

for grupo in grupos:
    sub_df = df_clean[df_clean['SEXO'] == grupo]
    
    for col in cols:
        datos = sub_df[col].dropna()
        estadistico, p_value = stats.shapiro(datos)
        
        conclusion = (
            f"Los datos de {col.lower()} en {grupo} no siguen una distribución normal"
            if p_value < 0.05
            else f"Los datos de {col.lower()} en {grupo} sí siguen una distribución normal"
        )
        
        print(f"Shapiro-Wilk para {col} en sexo {grupo}: W={estadistico}, p-value={p_value}")
        print("Conclusión:", conclusion)
        print()
```

```python
## 1.2 Visualizar distribuciones con QQ plots e histogramas
cols = ['INGRESOS', 'EGRESOS']
grupos = list(df_clean['SEXO'].unique())

# QQ plots 
fig, axes = plt.subplots(
    nrows=len(grupos),
    ncols=len(cols),
    figsize=(12, 8)
)

for i, grupo in enumerate(grupos):
    sub_df = df_clean[df_clean['SEXO'] == grupo]
    for j, col in enumerate(cols):
        ax = axes[i, j] if len(grupos) > 1 else axes[j]
        sm.qqplot(sub_df[col].dropna(), line='s', ax=ax)
        ax.set_title(f"QQ Plot - {col} ({grupo})")

plt.tight_layout()
plt.show()

# Histogramas 
fig, axes = plt.subplots(
    nrows=len(grupos),
    ncols=len(cols),
    figsize=(12, 8)
)

for i, grupo in enumerate(grupos):
    sub_df = df_clean[df_clean['SEXO'] == grupo]
    for j, col in enumerate(cols):
        ax = axes[i, j] if len(grupos) > 1 else axes[j]
        sns.histplot(sub_df[col].dropna(), kde=True, ax=ax)
        ax.set_title(f"Histograma - {col} ({grupo})")

plt.tight_layout()
plt.show()
```

```python
## 2. Pruebas no paramétricas ya que no se cumple requisito de normalidad en los ingresos y egresos

## 2.1 Prueba de Mann-Whitney U para comparar ingresos entre hombres y mujeres
ingresos_hombres = df_clean[df_clean['SEXO'] == 'M']['INGRESOS']
ingresos_mujeres = df_clean[df_clean['SEXO'] == 'F']['INGRESOS']
u_stat, p_value = stats.mannwhitneyu(ingresos_hombres, ingresos_mujeres, alternative='two-sided')
print(f"Mann-Whitney U: U={u_stat}, p-value={p_value}")
print("Conclusión:", "La diferencia en ingresos entre hombres y mujeres es estadisticamente significativa" if p_value < 0.05 else "No hay diferencia estadisticamente significativa en ingresos entre hombres y mujeres")

# 2.2 Prueba de Mann-Whitney U para comparar egresos entre hombres y mujeres
egresos_hombres = df_clean[df_clean['SEXO'] == 'M']['EGRESOS']
egresos_mujeres = df_clean[df_clean['SEXO'] == 'F']['EGRESOS']
u_stat, p_value = stats.mannwhitneyu(egresos_hombres, egresos_mujeres, alternative='two-sided')
print(f"Mann-Whitney U: U={u_stat}, p-value={p_value}")
print("Conclusión:", "La diferencia en egresos entre hombres y mujeres es estadisticamente significativa" if p_value < 0.05 else "No hay diferencia estadisticamente significativa en egresos entre hombres y mujeres")
```

Tras filtrar las transacciones monetarias (VALOR > 0), se aplicó la prueba de Shapiro-Wilk, demostrando que los datos no siguen una distribución normal. Por lo tanto, se utilizó la prueba no paramétrica U de Mann-Whitney (Wilcoxon para muestras independientes). Los resultados arrojaron valores p cercanos a cero en ambos casos, confirmando que sí existe una diferencia estadísticamente significativa tanto en ingresos como en egresos entre hombres y mujeres. Al evaluar las medianas para evitar el sesgo de los valores atípicos, se concluye que los hombres ingresan más dinero (8,008,000 vs 5,000,000) y también gastan más (2,500,000 vs 2,000,000).
