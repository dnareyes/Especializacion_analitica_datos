## 2. Análisis univariado de variables cualitativas.

Seleccione al menos tres variables categóricas y construya:

- Tabla de frecuencias absolutas, relativas, acumuladas y porcentuales:


```python
cols = ['Income_type', 'Education_type', 'Family_status']

for n in cols:
    print(f"\n--- Tabla de frecuencias: {n} ---")
    tab = df[n].value_counts().to_frame(name='Cant.')
    tab['%'] = (df[n].value_counts(normalize=True) * 100).round(2)
    tab['Cant. Acum.'] = tab['Cant.'].cumsum()
    tab['% Acum.'] = tab['%'].cumsum().round(2)
    
    print(tab)
```

En este análisis de variables categoricas podemos observar que predomina la fuerza laboral, los nivbeles educativos medios (educación secundaria) y el estado civil de las personas casadas.

1. `Income_type`: Teniendo en cuenta que las categorais working, comercial associate y state servant pueden clasificarse bajo una sola, podemos decir que el 82.34% de las personas reciben ingresos de un trabajo. Mientras que 17.63 son pensionados y el 0.03% restante estudia.

2. `Education_type`: El nivel educativo se concentra en la educacion media, solo 1 de 4 personas aproximadamente tiene educación superior, y el 0.06% restante tiene educación académica.
   
3. `Family_status`: El estado civil casdo es el más comun con un 67.26% de los casos.

El perfil dominante en este dataset es de un individuo trabajador en el sector privado, con educacion secundaria terminada y casado.

```python
fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(22, 7))

income_counts = df['Income_type'].value_counts()
education_counts = df['Education_type'].value_counts()
family_counts = df['Family_status'].value_counts()

ax[0].barh(income_counts.index, income_counts.values, color='skyblue', edgecolor='black', alpha=0.7)
ax[0].set_title('Distribution of Income Type', fontweight='bold')  
ax[0].set_xlabel('Income Type')
ax[0].set_ylabel('Frequency')

ax[1].barh(education_counts.index, education_counts.values, color='salmon', edgecolor='black', alpha=0.7)
ax[1].set_title('Distribution of Education Type', fontweight='bold')
ax[1].set_xlabel('Education Type')
ax[1].set_ylabel('Frequency')

ax[2].barh(family_counts.index, family_counts.values, color='lightgreen', edgecolor='black', alpha=0.7)
ax[2].set_title('Distribution of Family Status', fontweight='bold')
ax[2].set_xlabel('Family Status')
ax[2].set_ylabel('Frequency')
ax[2].color = 'lightgreen'

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(22, 7))

ax[0].pie(income_counts.values, labels=income_counts.index, autopct='%1.2f%%')
ax[0].set_title('Distribution of Income Type', fontweight='bold')  

ax[1].pie(education_counts.values, labels=education_counts.index, autopct='%1.2f%%')
ax[1].set_title('Distribution of Education Type', fontweight='bold')

ax[2].pie(family_counts.values, labels=family_counts.index, autopct='%1.2f%%')
ax[2].set_title('Distribution of Family Status', fontweight='bold')

plt.tight_layout()
plt.show()
```

Si bien el gráfico de pastel es visualmente atractivo para resaltar una parte proporcional de un todo y es fácil de entender, creo que no es la mejor opcion para represetnar como se distribuyen estas variables, si lo comparamos con el grafico de barras, en este se observan diferencias más claras entre las categorias.
