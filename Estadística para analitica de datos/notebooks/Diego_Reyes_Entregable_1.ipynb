## 4. Tablas de frecuencias para variables numéricas.
Selecciona dos variables cuantitativas y:

- Determine el numero de clases usando la regla de Sturges:

```python
n = len(df['Total_income'])
k = int(np.ceil(1 + 3.322 * np.log10(n)))
print(f"Número de clases: {k}")
```

- Tabla de frecuencias absolutas, relativas y acumuladas:

```python
cols = ['Total_income', 'Years_employed']

for n in cols:

    print(f"\n--- Tabla de frecuencias: {n} ---")
    k = int(np.ceil(1 + 3.322 * np.log10(len(df[n]))))
    intervalos = pd.cut(df[n], bins=k)
    tab = intervalos.value_counts().sort_index().to_frame(name='Cant.')
    tab['%'] = (intervalos.value_counts(normalize=True).sort_index() * 100).round(2)
    tab['Cant. Acum.'] = tab['Cant.'].cumsum()
    tab['% Acum.'] = tab['%'].cumsum().round(2)
    tab['Marca de Clase (Xi)'] = [i.mid for i in tab.index]
    
    display(tab)
```

1. `Total_income`: El número de clases según la regla de Sturges es de 15. La distribución del ingreso total se concentra en los dos primeros itnervalos ya que el 80.64 % de las personas ganan entre 25,452 y 233,400. La distribución es segada a la izquierda (asimetría positiva), indicando que la mayoría de las personas tienen ingresos bajos, mientras que un pequeño porcentaje tiene ingresos muy altos. Hay un valor atípico en el intervalo de 955,800 a 1,059,000 que representa el 0.01% de los casos.

2. `Years_employed`: El número de clases según la regla de Sturges es de 15. En este caso el 42.63% de las personas tienen menos de 2.8 años en su empleo actual. Si sumamos el segundo intervalo, descubrimos que el 63.51% de las personas tienen menos de 5.7 años en su empleo actual. Hay un decenso progresivo a medida que aumentan los intervalos de años trabajados.

En general para estas dos variables podemos decir que la mayoría de las personas tienen ingresos bajos y poca experiencia laboral.

## 6. Medidas de dispersión y variablilidad:

Para las mismas variables del punto anterior, calcule:

```python
cols = ['Total_income', 'Years_employed', 'Age']

summary = {
    'Rango': df[cols].max() - df[cols].min(),
    'Varianza': df[cols].var(),
    'Desviación estándar': df[cols].std(),
    'Coeficiente de variación': df[cols].std() / df[cols].mean() * 100,
    'Rango intercuartílico': df[cols].quantile(0.75) - df[cols].quantile(0.25),
}

deviation_table = pd.DataFrame(summary).round(2)
display(deviation_table)
```

## 9. Medidas de forma y curtosis:

Para al menso dos variables numéricas, calcule:

- Coeficiente de asimetría (skewness):

```python
cols = ['Total_income', 'Years_employed', 'Age']

for i in cols:
    skewness = scipy.stats.skew(df[[i]], axis=0, bias=True)
    print(f"Coeficiente de asimetría para {i}: {skewness}")

for i in cols:
    kurtosis = scipy.stats.kurtosis(df[[i]], axis=0, bias=True)
    print(f"Curtosis para {i}: {kurtosis}")
```

1. `Total_Income`: Teniendo en cuenta el coficiente de asimetria se puede decir que la distribución es asimetrica hacia la derecha (asimetría positiva) ya que su valor es de 2.65. Indicando que la mayoria de ingresos se concentran en los valores más bajos, para el caso de la kurtosis, el valor es de 15.77, lo que indica una distribución leptocúrtica, con colas más pesadas y un pico más alto que una distribución normal.

2. `Years_employed`: El coeficiente de asimetría de 1.84 también indica una distribución asimétrica hacia la derecha, aunque menos pronunciada que en el caso de los ingresos. La kurtosis de 4.22 sugiere una distribución leptocúrtica, con colas más pesadas y un pico más alto que una distribución normal, aunque menos extrema que en el caso de los ingresos.

3. `Age`: El coeficiente de asimetría de 0.15 indica una distribución casi simétrica, cercana a la distribución normal ya que se acerca a 0, y la kurtosis de -1.05 indica una distribución platicúrtica, con colas más ligeras y un pico más bajo que una distribución normal.

## 10. Detección y análisis de valores atípicos:

Utilizando diagramas de caja y criterios estadisticos:

- Identifique posibles valores atípicos
- Explique como afectan las medidas de tendencia central y dispersión
- Compare resultados antes y despues de eliminar outliers

```python
sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(25, 12))

sns.boxplot(data=df, x='Total_income', color='skyblue', ax=axes[0])
sns.boxplot(data=df, x='Years_employed', color='salmon', ax=axes[1])
sns.boxplot(data=df, x='Age', color='lightgreen', ax=axes[2])

plt.tight_layout()
plt.show()
```

Usando los diagramas de cajas para las variables `Total_income` y `Years_employed`, se identifican varios valores atípicos que se encuentran por encima del rango intercuartílico. Estos outliers representan casos extremos de ingresos muy altos a partir de aproximadamente 400.000 dólares y casos de tiempo de empleo muy amplios a partir de aproximadamente 20. Para la variable `Age`, no se identifican valores atípicos, ya que los datos se distribuyen de forma relativamente uniforme entre los diferentes rangos de edad.

```python
iqr_income = stats.iqr(df['Total_income'])
print(f"Rango intercuartílico para Total_income: {iqr_income}")
q1_income = df['Total_income'].quantile(0.25)
q3_income = df['Total_income'].quantile(0.75)
income_upper_bound = q3_income + 1.5 * iqr_income
income_lower_bound = q1_income - 1.5 * iqr_income

iqr_years = stats.iqr(df['Years_employed'])
print(f"Rango intercuartílico para Years_employed: {iqr_years}")
q1_years = df['Years_employed'].quantile(0.25)
q3_years = df['Years_employed'].quantile(0.75)
years_upper_bound = q3_years + 1.5 * iqr_years
years_lower_bound = q1_years - 1.5 * iqr_years

iqr_age = stats.iqr(df['Age'])
print(f"Rango intercuartílico para Age: {iqr_age}")
q1_age = df['Age'].quantile(0.25)
q3_age = df['Age'].quantile(0.75)
age_upper_bound = q3_age + 1.5 * iqr_age
age_lower_bound = q1_age - 1.5 * iqr_age

 
df_clean = df[
    (df['Total_income'] >= income_lower_bound) & (df['Total_income'] <= income_upper_bound) &
    (df['Years_employed'] >= years_lower_bound) & (df['Years_employed'] <= years_upper_bound) &
    (df['Age'] >= age_lower_bound) & (df['Age'] <= age_upper_bound)
]

print(f"\nObservaciones originales: {len(df)}")
print(f"Observaciones después de eliminar outliers: {len(df_clean)}")
print(f"Outliers eliminados: {len(df) - len(df_clean)}")
```
