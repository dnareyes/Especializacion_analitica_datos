
# Modelo Estrella

En analítica de datos es común organizar la información en un **modelo dimensional**.

- **Tabla de hechos:** contiene las métricas del negocio.
- **Tablas de dimensión:** contienen el contexto de las métricas.

En este ejercicio:

**fact_ventas**
- id_tienda
- id_producto
- fecha
- cantidad
- precio

**dim_tienda**
- id_tienda
- nombre_tienda
- ciudad

**dim_producto**
- id_producto
- nombre_producto
- categoria
