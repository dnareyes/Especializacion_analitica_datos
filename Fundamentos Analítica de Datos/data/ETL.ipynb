```python
import pandas as pd
import numpy as np

df_full = pd.read_csv("Fundamentos Analítica de Datos\Datos\fact_ventas.csv")

# 39 filas que no tengan nulos en Metodo_Pago (para inyectarlos luego tú)
df_unicos = df_full[df_full["Metodo_Pago"].notna()].iloc[10:49].copy()
print(len(df_unicos))  # → 39

# Tomamos la fila en el índice 5 (cualquier fila sirve)
fila_a_duplicar = df_full.iloc[[5]].copy()

# La repetimos 11 veces
duplicados = pd.concat([fila_a_duplicar] * 11, ignore_index=True)
print(len(duplicados))  # → 11
print(duplicados.duplicated().sum())  # → 10 duplicados detectados

df_50 = pd.concat([df_unicos, duplicados], ignore_index=True)
print(len(df_50))                    # → 50
print(df_50.duplicated().sum())      # → 10 (los 10 extras de los 11)
```
