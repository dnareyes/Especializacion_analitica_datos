# Calidad de Datos en Python con Pandas

## Objetivo
Detectar y corregir problemas de calidad de datos relacionados con:
- claves inexistentes
- valores nulos
- integridad referencial


```python
import pandas as pd
import numpy as np
```

## Crear tabla de dispositivos

```python
dispositivos = pd.DataFrame({
'id_dispositivo':[101,102,103],
'nombre':['Laptop','Tablet','Consola'],
'tipo_dispositivo':['Computador','Tablet','Gaming'],
'precio':[3500,1200,2200]
})

dispositivos
```

## Crear tabla de préstamos con errores

```python
prestamos = pd.DataFrame({
'id_prestamo':[1,2,3,4,5],
'id_dispositivo':[101,102,999,103,np.nan],
'pais':['Colombia','Mexico','Colombia','Colombia','Mexico']
})

prestamos
```

## Detectar dispositivos inexistentes (Inconsistencias)

```python
invalidos = prestamos[~prestamos['id_dispositivo'].isin(dispositivos['id_dispositivo'])]
invalidos
```

## Detectar valores nulos

```python
nulos = prestamos[prestamos['id_dispositivo'].isna()]
nulos
```

```python
total_nulos = prestamos['id_dispositivo'].isna()

total_nulos
```

## Sumar valores nulos

```python
total_nulos = prestamos['id_dispositivo'].isna().sum()
total_nulos
```

```python
total_nulos = prestamos.isna().sum()
total_nulos
```

```python
prestamos.isna()
```

## Reemplazar nulos con valores diferentes por columna

```python
prestamos
```

```python
prestamos = prestamos.fillna({
"id_dispositivo":0,
"pais_prestamo":"Desconocido",
"precio":0
})
```

```python
prestamos
```

## Reemplazar valores incorrectos

```python
# convertir a tipo numerico
prestamos['id_dispositivo'] = pd.to_numeric(prestamos['id_dispositivo'], errors='coerce')
```

```python
prestamos['id_dispositivo'] = prestamos['id_dispositivo'].replace(999,0)


prestamos
```

```python
prestamos['id_dispositivo'] = prestamos['id_dispositivo'].fillna(0)
prestamos
```

## Integrar datasets

```python
dataset = prestamos.merge(dispositivos, on='id_dispositivo', how='left')
dataset
```
