```python
# Importar pandas
import pandas as pd
```

El concepto de datos ordenados desempeña un papel crucial en este proceso. Los principios de datos ordenados abogan por una estructura donde **cada variable forma una columna **y **cada observación una fila.** Esta estructura consistente facilita una manipulación, análisis y visualización de datos más sencillas y eficaces.




Captura de pantalla 2026-03-16 161857.png



Para análisis de datos, el formato más recomendable generalmente es el DataFrame en formato largo (long / tidy).

1. Cada variable debe tener su propia columna
2. Cada observación debe tener su propia fila
3. Cada valor debe tener su propia celda

# **melt()** en pandas se usa para transformar un DataFrame de ancho a largo.

Parámetros principales

**id_vars:** columnas que se conservan fijas

**var_name:** nombre de la nueva columna que contendrá los nombres de las columnas derretidas

**value_name:** nombre de la nueva columna con los valores

```python
archivo=pd.read_excel("/content/Datos.xlsx", sheet_name="Hospitales")
```

```python
archivo.head(3)
```

```python
# Renombrar columnas

archivo = archivo.rename(columns={'Costo diario GRD1 [MM$]': 'GRD1','Costo diario GRD2 [MM$]':'GRD2','Costo diario GRD3 [MM$]':'GRD3','Costo diario GRD4 [MM$]':'GRD4','Costo diario GRD5 [MM$]':'GRD5','Costo diario GRD6 [MM$]':'GRD6','Costo diario GRD7 [MM$]':'GRD7','Costo diario GRD8 [MM$]':'GRD8','Costo diario GRD9 [MM$]':'GRD9','Costo diario GRD10 [MM$]':'GRD10'})
```

```python
archivo=archivo.rename(columns={'Nombre Servicio de Salud':'Servicio'})
```

```python
archivo.head(3)
```

```python
# subconjunto
nuevo_archivo = archivo[["Servicio", "GRD1", "GRD2","GRD3","GRD4","GRD5","GRD6","GRD7","GRD8","GRD9","GRD10"]]
```

```python
nuevo_archivo.head(3)
```

# Función **MELT()**

```python
archivo_tidy = nuevo_archivo.melt(
    id_vars="Servicio",
    var_name="GRD",
    value_name="Costo_Diario"
)
```

```python
archivo_tidy.info()
```

```python
archivo_tidy.head(25)
```

**Calcular promedio de costos por dignostico**

```python

archivo_tidy.groupby("GRD")["Costo_Diario"].mean().sort_values(ascending=False)
```

```python
archivo2=pd.read_excel("/content/Datos.xlsx", sheet_name="Información Histórica")
```

**Calcular los promedios históricos de estadía por Hospital**

```python

archivo2.groupby("Hospital al que llega")["Estadia"].mean().round(2).sort_values(ascending=False)
```
