## Importar librerías

```python

import pandas as pd
import numpy as np
```

## Crear datasets con datos sucios

```python

dispositivos = pd.DataFrame({
"id_dispositivo":[101,102,103,103],
"nombre":["Laptop","Tablet","Consola","Consola"],
"tipo_dispositivo":["Computacion","Comput@cion","Computacion ","Computacion"]
})

dispositivos
```

```python

prestamos = pd.DataFrame({
"id_prestamo":[1,2,3,4,5],
"id_dispositivo":[101,102,999,103,101],
"pais_prestamo":["Colombia ","Mexic","USA","Colombia","Colombia"],
"moneda_operacion":["COP","MXN","USD","USD","COP"],
"prestamo_virtual":["True","1","Si","No","true"]
})

prestamos
```

```python

paises = pd.DataFrame({
"pais":["Colombia","Mexico","USA"],
"moneda_oficial":["COP","MXN","USD"]
})

paises
```

```python
dataset_sucio = prestamos.merge(dispositivos,on="id_dispositivo",how="left")
dataset_sucio = dataset_sucio.merge(paises,left_on="pais_prestamo",right_on="pais",how="left")

dataset_sucio
```


# Ejercicios de Limpieza de Datos

Corregir los siguientes **10 problemas**:

1. Espacios en **pais_prestamo**
2. Error tipográfico en **Mexic**
3. Categoría **Comput@cion**
4. Espacio en **Computacion **
5. Registros duplicados en **dispositivos**
6. ID inexistente **999**
7. Valores booleanos inconsistentes
8. Texto **true** en minúscula
9. Moneda incorrecta para Colombia
10. Integración de datasets



## Ejercicio 1 — Eliminar espacios en texto

```python

prestamos["pais_prestamo"] = prestamos["pais_prestamo"].str.strip()
```

## Ejercicio 2 — Corregir país mal escrito

```python

prestamos["pais_prestamo"] = prestamos["pais_prestamo"].replace({
"Mexic":"Mexico"
})
```

## Ejercicio 3 — Limpiar categorías

```python

dispositivos["tipo_dispositivo"] = dispositivos["tipo_dispositivo"].str.strip()

dispositivos["tipo_dispositivo"] = dispositivos["tipo_dispositivo"].replace({
"Comput@cion":"Computacion"
})
```

## Ejercicio 4 — Eliminar duplicados

```python

dispositivos = dispositivos.drop_duplicates()
```

## Ejercicio 5 — Detectar dispositivos inexistentes

```python

invalidos = prestamos[~prestamos["id_dispositivo"].isin(dispositivos["id_dispositivo"])]

invalidos
```

## Ejercicio 6 — Normalizar booleanos

```python

prestamos["prestamo_virtual"] = prestamos["prestamo_virtual"].replace({
"True":True,
"true":True,
"1":True,
"Si":True,
"No":False
})
```

## Ejercicio 7 — Integrar datasets

```python

dataset = prestamos.merge(dispositivos,on="id_dispositivo",how="left")
dataset = dataset.merge(paises,left_on="pais_prestamo",right_on="pais",how="left")

dataset
```

## Ejercicio 8 — Auditoría de moneda

```python

errores_moneda = dataset[dataset["moneda_operacion"] != dataset["moneda_oficial"]]

errores_moneda
```

```python
# corregir
# opcion 1 En Pandas, loc es un método de selección y modificación de datos basado en etiquetas (labels). Se utiliza para acceder o modificar filas y columnas usando el nombre de las columnas o el índice.
dataset.loc[
    dataset["moneda_operacion"] != dataset["moneda_oficial"],
    "moneda_operacion"
] = dataset["moneda_oficial"]

dataset
```

```python
# opcion 2. si la moneda es correcta → mantener
# si es incorrecta → reemplazar por moneda_oficial

dataset["moneda_operacion"] = dataset["moneda_operacion"].where(
    dataset["moneda_operacion"] == dataset["moneda_oficial"],
    dataset["moneda_oficial"]
)

dataset
```
